```python
def normalize(url):
    if not url.startswith (("http://", "https://")):
        url = "http://" + url
    return url
```

```python
df1['domain_normal'] = df1['domain'].apply(normalize)
```

```python
url = df1["domain_normal"].iloc[0]
print(url)
```
